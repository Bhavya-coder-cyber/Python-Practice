```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("Salary.csv")
```

```python
x = data[['YearsExperience']]
y = data[['Salary']]
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
data['PredictedSalaries'] = model.predict(x)
```

```python
print(f"{round(model.coef_[0][0])}")
print(f"{round(model.intercept_[0])}")
```

```python
plt.scatter(x,y,color="blue",label="Actual Data")
plt.plot(x,data['PredictedSalaries'],color="red",label="Regression Line")
plt.xlabel("Years Of Experience")
plt.ylabel("Salary")
plt.title("Salary VS Experience")
plt.legend()
plt.grid(True)
```
