```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x=np.arange(1,10)
y=np.arange(11,20)
```

```python
plt.scatter(x,y,c='b')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Demo Graph')
plt.savefig('Demo.png')
```

```python
plt.subplot(2,2,1)
plt.plot(x,y,'b')
plt.subplot(2,2,2)
plt.plot(x,y,'r*--')
plt.subplot(2,2,3)
plt.plot(x,y,'yo')
```

```python
data = [np.random.normal(0,std,100) for std in range(1,4)]
plt.boxplot(data,patch_artist=True)
```

```python
labels = 'Python', 'C++', 'Java', 'JavaScript'
Sizes = [100,200,300,400]
colors = ['red','yellow','green','blue']
explode = (0, 0.2, 0, 0)

plt.pie(Sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True)
plt.axis('equal')
plt.show()
```

```python
import seaborn as sns #Most important in exploratory data analysis
```

```python
df = sns.load_dataset("tips")
```

```python
df.head() #Tip is the only dependent variable whereas others are independent variables
```

## Coefficient Correlation

A correlation heatmap uses colored cells, typically in a monochromatic scale, to show a 2D correlation matrix (table) between
two discrete dimensions or event types. It is very important in Feature Selection. The values should be either int or float.

```python
df.corr(numeric_only=True)
```

```python
sns.heatmap(df.corr(numeric_only=True)) #This is useful in Feature analysis
```

```python
sns.jointplot(x='tip',y='total_bill',data=df,kind='hex')
```

```python
sns.jointplot(x='tip',y='total_bill',data=df,kind='reg')
```

```python
#Using pairplot for more than two features.
sns.pairplot(df)
```

```python
sns.pairplot(df, hue='sex')
```

```python
sns.distplot(df['tip'])
```

## Categorical Plots

```python
sns.countplot(data=df, x='smoker', palette="Set2")
```

```python
sns.countplot(data=df, x='day', palette="Set1")
```

```python
sns.boxplot(x='smoker',y='total_bill',data=df,palette="Set1")
```

```python
sns.boxplot(x='day',y='total_bill',hue='sex',data=df,palette="rainbow")
```
